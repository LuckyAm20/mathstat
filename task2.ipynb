## Задание 1. Партия обуви
Компания утверждает, что длина шнурков в партии обуви имеет среднее значение 120 см, и стандартное отклонение в партии точно известно и составляет 5 см. Из выборки 25 шнурков было получено среднее значение длины 122 см. Найдите 99%-ый доверительный интервал для среднего значения длины шнурков, границы округлите до целых с помощью функции int

(1 балл)

```python
import scipy.stats as st
import math

sigma = 5
n = 25
x_mean = 122
z = st.norm.ppf(1 - 0.01 / 2)

lower = x_mean - z * sigma / math.sqrt(n)
upper = x_mean + z * sigma / math.sqrt(n)

int(lower), int(upper)
```

##Задание 2. Среднее время заказа
Компания исследует среднее время выполнения заказа курьерской службой. Было случайным образом выбрано 10 заказов, для которых среднее время доставки составило 45 минут, а выборочная дисперсия — 9 минут. Постройте 95%-ый доверительный интервал для среднего времени доставки заказа
(1 балл)

```python
import scipy.stats as st
import math

n = 10
x_mean = 45
s = 3
t = st.t.ppf(1 - 0.05 / 2, df=n - 1)

lower = x_mean - t * s / math.sqrt(n)
upper = x_mean + t * s / math.sqrt(n)

round(lower, 2), round(upper, 2)
```

## Задание 3. Повторные покупки
Интернет-магазин хочет определить долю пользователей, которые возвращаются для повторной покупки в течение месяца после первой покупки. Для этого он проанализировал выборку из 800 клиентов, и оказалось, что 240 из них сделали повторную покупку в течение месяца. Необходимо определить 95% доверительный интервал для истинной доли клиентов, которые возвращаются для повторной покупки в течение этого времени

(1 балл)


```python
import scipy.stats as st
import math

n = 800
p_hat = 240 / 800
z = st.norm.ppf(1 - 0.05 / 2)

lower = p_hat - z * math.sqrt(p_hat * (1 - p_hat) / n)
upper = p_hat + z * math.sqrt(p_hat * (1 - p_hat) / n)

round(lower, 2), round(upper, 2)
```

## Задание 4. Аниме, средний рейтинг

Постройте 95% доверительный интервал для среднего рейтинга по всем аниме

(1 балл)

```python
import pandas as pd
import scipy.stats as st
import numpy as np

df_anime = pd.read_csv('anime.csv')

ratings = df_anime['rating'].dropna()
mean = ratings.mean()
se = ratings.std(ddof=1) / np.sqrt(len(ratings))
ci = st.t.interval(0.95, df=len(ratings)-1, loc=mean, scale=se)
ci
```

(1 балл)

```python
comedy = df_anime[df_anime['genre'].str.contains('Comedy', na=False)]['rating'].dropna()
mean = comedy.mean()
se = comedy.std(ddof=1) / np.sqrt(len(comedy))
ci = st.t.interval(0.95, df=len(comedy)-1, loc=mean, scale=se)
ci
```

## Задание 6. Средние рейтинги по типам аниме

## Задание 7. Самый широкий доверительный интервал по типам

Определите, для какого жанра, исключая nan (none), аниме 95% доверительный интервал самый широкий

(1 балл)

```python
result = {}

for t, group in df_anime.groupby('type'):
    r = group['rating'].dropna()
    if len(r) > 1:
        se = r.std(ddof=1) / np.sqrt(len(r))
        ci = st.t.interval(0.95, df=len(r)-1, loc=r.mean(), scale=se)
        result[t] = ci[1] - ci[0]

max_type = max(result, key=result.get)
max_type, result[max_type]
```

## Задание 8. Самый широкий доверительный интервал по жанрам

Определите для какого жанра аниме 95% доверительный интервал самый широкий

(1 балл)

```python
genres = {}

for g in df_anime['genre'].dropna().unique():
    for subg in [x.strip() for x in g.split(',')]:
        sub = df_anime[df_anime['genre'].str.contains(subg, na=False)]['rating'].dropna()
        if len(sub) > 1:
            se = sub.std(ddof=1) / np.sqrt(len(sub))
            ci = st.t.interval(0.95, df=len(sub)-1, loc=sub.mean(), scale=se)
            genres[subg] = ci[1] - ci[0]

max_genre = max(genres, key=genres.get)
max_genre, genres[max_genre]
```

## Задание 9. Самая длинная медианная сезонность

Найдите жанр, у которого максимальная медиана по кол-ву эпизодов

(1 балл)

```python
genre_median = {}

for g in df_anime['genre'].dropna().unique():
    for subg in [x.strip() for x in g.split(',')]:
        sub = df_anime[df_anime['genre'].str.contains(subg, na=False)]['episodes'].replace('Unknown', np.nan).dropna().astype(float)
        if len(sub) > 0:
            genre_median[subg] = sub.median()

max_genre = max(genre_median, key=genre_median.get)
max_genre, genre_median[max_genre]
```

## Задание 10. Самый узкий ДИ

Найдите жанр, у которого самый узкий 95% доверительный интервал по подписчикам

## Задание 13. Средний чек по категориям

Найдите категорию с самым широким 95% доверительным интервалом по среднему чеку (quantity * price)

(1 балл)

```python
result = {}

for cat, group in df.groupby('category'):
    r = group['check']
    if len(r) > 1:
        se = r.std(ddof=1) / np.sqrt(len(r))
        ci = st.t.interval(0.95, df=len(r)-1, loc=r.mean(), scale=se)
        result[cat] = ci[1] - ci[0]

max_cat = max(result, key=result.get)
max_cat, round(result[max_cat], 2)
```

## Задание 17. Разница в средних чеках

Постройте 95% доверительный интервал для разницы по среднему чеку (quantity * price) для мужчин и женщин, результат округлите до двух знаков после запятой

(1 балл)

```python
male = df[df['gender'] == 'Male']['check']
female = df[df['gender'] == 'Female']['check']

diff = male.mean() - female.mean()
# стандартная ошибка разности средних
se = np.sqrt(male.var(ddof=1)/len(male) + female.var(ddof=1)/len(female))
dfree = min(len(male), len(female)) - 1
ci = st.t.interval(0.95, df=dfree, loc=diff, scale=se)
tuple(round(x, 2) for x in ci)
```

## Задание 18. Средний чек по месяцам

Определите месяц с наибольшей шириной 95% доверительного интервала среднего чека

(1 балл)

```python
df['invoice_date'] = pd.to_datetime(df['invoice_date'])
df['year'] = df['invoice_date'].dt.year
df['month'] = df['invoice_date'].dt.month

widths = {}

for (y, m), g in df.groupby(['year', 'month']):
    r = g['check']
    if len(r) > 1:
        se = r.std(ddof=1) / np.sqrt(len(r))
        ci = st.t.interval(0.95, df=len(r)-1, loc=r.mean(), scale=se)
        widths[(y, m)] = ci[1] - ci[0]

min_year_month = min(widths, key=widths.get)
min_year_month, round(widths[min_year_month], 2)

```

## Задание 19. Возрастная группа с максимальными тратами

Разбейте пользователей по возрасту на группы: <20, 20–40, 40-60, >60
Определите, у какой группы наибольшая верхняя граница 95% доверительного интервала по среднему чеку

(1 балл)

```python
bins = [0, 20, 40, 60, 100]
labels = ['<20', '20–40', '40–60', '>60']
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=False)

group_ci = {}

for group, g in df.groupby('age_group'):
    r = g['check']
    if len(r) > 1:
        se = r.std(ddof=1) / np.sqrt(len(r))
        ci = st.t.interval(0.95, df=len(r)-1, loc=r.mean(), scale=se)
        group_ci[group] = ci[1]

max_group = max(group_ci, key=group_ci.get)
max_group, round(group_ci[max_group], 2)
```
